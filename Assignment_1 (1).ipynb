```python
# Creating a student dictionary with 5 elements
my_dict = {
    'name': 'Rahul',
    'roll_no': 101,
    'age': 20,
    'city': 'Pune',
    'grade': 'A',
}

print('Initial Dictionary:', my_dict)

# Adding information
my_dict['email'] = 'rahul@email.com'

# Delet information
del my_dict['age']

print('Dictionary after add and delete:', my_dict)
```

```python
my_dict = {'name': 'ABC', 'panel': 'B', 'rollno': 34, 'marks': [65, 87, 67, 94]}

keys_list = list(my_dict.keys())
values_list = list(my_dict.values())

print('Keys List:', keys_list)
print('Values List:', values_list)
```

```python
mydict = {
    'marks1': 23,
    'marks2': 123,
    'marks3': 43,
    'marks4': 13,
    'marks5': 39,
}

values = list(mydict.values())
mean_val = sum(values) / len(values)

print('Mean of all values:', mean_val)
```

```python
my_dict = {
    'name': ['Yash', 'Neel', 'Dev'],
    'rollno': [11, 12, 13],
    'marks': [78, 56, 98],
}

# 1. Display name as Dev
dev_index = my_dict['name'].index('Dev')
print('a) Name:', my_dict['name'][dev_index])

# b) Display 12 roll no
roll_index = my_dict['rollno'].index(12)
print('b) Roll no:', my_dict['rollno'][roll_index])

# c) Display greatest marks
print('c) Greatest marks:', max(my_dict['marks']))
```

```python
text = input('Enter a string: ')
frequency = {}

for char in text:
  if char in frequency:
    frequency[char] += 1
  else:
    frequency[char] = 1

print('Character Frequency:', frequency)
```

```python
numerator = int(input('Enter numerator: '))
denominator = int(input('Enter denominator: '))

quotient = numerator // denominator
remainder = numerator % denominator

result_tuple = (quotient, remainder)
print('Tuple of (Quotient, Remainder):', result_tuple)
```

```python
from datetime import date

print('Enter first date:')
d1 = int(input('Day: '))
m1 = int(input('Month: '))
y1 = int(input('Year: '))
date1_tuple = (d1, m1, y1)

print('Enter second date:')
d2 = int(input('Day: '))
m2 = int(input('Month: '))
y2 = int(input('Year: '))
date2_tuple = (d2, m2, y2)

# Convert tuples to date objects
date1_obj = date(date1_tuple[2], date1_tuple[1], date1_tuple[0])
date2_obj = date(date2_tuple[2], date2_tuple[1], date2_tuple[0])

# Find difference in days
days_diff = abs((date2_obj - date1_obj).days)
print('Number of days between the two dates:', days_diff)
```
