```python
marks = int(input("Enter marks: "))

if marks >= 90:
    print("Grade = O")

elif marks >= 80:
    print("Grade = A+")

elif marks >= 70:
    print("Grade = A")

elif marks >= 60:
    print("Grade = B")

elif marks >= 50:
    print("Grade = C")

elif marks >= 40:
    print("Grade = P")

else:
    print("Grade = F")
```

```python
mydict = {'marks1':23, 'marks2':123, 'marks3':43, 'marks4':13, 'marks5':39}


total = 0

for i in mydict.values():
    total = total + i

mean = total / len(mydict)

print("Mean of all values =", mean)
```

```python
#frequency of string

string = input("Enter a string: ")

for i in string:
    count = 0

    for j in string:
        if i == j:
            count = count + 1

    print(i, "=", count)
```

```python
numerator = int(input("Enter numerator: "))
denominator = int(input("Enter denominator: "))

quotient = numerator // denominator
remainder = numerator % denominator

answer = (quotient, remainder)

print("Answer =", answer)
```
